```python
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter, MaxNLocator
from scipy.ndimage import gaussian_filter1d

# ========= 1) 选定的两个数据文件（按需替换路径） =========
files = {
    ("Chronic", "Algae"): "./EC10/algae_EC10_POP_72.xlsx",   # 慢性：EC10
    ("Acute",   "Algae"): "./EC50/algae_EC50_POP_72.xlsx",   # 急性：EC50/LC50
}

# 可能的数值列名（按优先级）
CANDS = ["mgperL","pred_mgperL","concentration","Concentration",
         "conc","Conc","toxicity","Toxicity","EC10","EC50","value","Value"]

def pick_col(df):
    for c in CANDS:
        if c in df.columns: 
            return c
    # 兜底：首个数值列
    for c in df.columns:
        if pd.api.types.is_numeric_dtype(df[c]): 
            return c
    raise ValueError("No numeric column found for concentration.")

def load_vals(path):
    if not os.path.exists(path):
        raise FileNotFoundError(path)
    df = pd.read_excel(path)
    col = pick_col(df)
    x = pd.to_numeric(df[col], errors="coerce").dropna().values
    x = x[np.isfinite(x)]
    return x[x >= 0]

# 读取
series = {(phase, tag): load_vals(p) for (phase, tag), p in files.items()}

# ========= 2) 公共对数分箱 =========
all_vals = np.concatenate([v for v in series.values() if len(v) > 0])
xmin = max(all_vals.min(), 1e-8)
xmax = np.percentile(all_vals, 99)   # 去掉极端尾
BINS = 60
bins = np.logspace(np.log10(xmin), np.log10(xmax), BINS + 1)
bins_log = np.log10(bins)
centers_log = 0.5 * (bins_log[:-1] + bins_log[1:])
widths_log = np.diff(bins_log)

# ========= 3) 视觉样式：Acute/Chronic 分开配色与线型 =========
phase_style = {
    "Acute":   dict(color="#1f77b4", linestyle="-"),   # 蓝 实线
    "Chronic": dict(color="#d62728", linestyle="--"),  # 红 虚线
}

# ========= 4) 绘图 =========
LABEL_FS, TICK_FS, LEGEND_FS, TITLE_FS = 18, 16, 15, 20
SMOOTH_SIGMA = 1.2  # 高斯平滑强度

plt.figure(figsize=(6.8, 4.6), dpi=500)
for (phase, tag), x in series.items():
    if len(x) == 0:
        continue
    counts, _ = np.histogram(x, bins=bins)
    N = counts.sum() if counts.sum() > 0 else 1.0

    # —— 每 log10 单位的概率密度（对 log 轴更直观）——
    density = counts.astype(float) / (N * widths_log)
    density = gaussian_filter1d(density, sigma=SMOOTH_SIGMA, mode="nearest")

    # —— 在 log 空间归一化：∫ density d(log10 x) = 1 —— 
    area = np.trapz(density, x=centers_log)
    if area > 0:
        density /= area

    st = phase_style[phase]
    # 阴影
    plt.fill_between(centers_log, 0, density, color=st["color"], alpha=0.15, label="_nolegend_")
    # 曲线
    plt.plot(centers_log, density, lw=2.6, color=st["color"], linestyle=st["linestyle"],
             label=f"{phase} ({tag})")

# x 轴使用线性坐标显示 log10 刻度为 10^k
ax = plt.gca()
ax.xaxis.set_major_locator(MaxNLocator(6))
ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: r"$10^{%d}$" % int(np.round(v))))

ax.set_xlabel("Concentration (mg/L)", fontsize=LABEL_FS)
ax.set_ylabel("Probability density", fontsize=LABEL_FS)
ax.tick_params(axis="both", which="major", labelsize=TICK_FS)
ax.tick_params(axis="both", which="minor", labelsize=TICK_FS - 2)
plt.grid(alpha=0.25)

plt.legend(frameon=False, ncol=1, loc="upper left",
           fontsize=LEGEND_FS, handlelength=2.8, handletextpad=0.6)

plt.tight_layout()
plt.show()

```
