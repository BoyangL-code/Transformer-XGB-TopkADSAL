```python
import numpy as np
import pickle

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score,root_mean_squared_error
from scipy.stats import pearsonr


def evaluate_regression_metrics(y_true, y_pred):
    """
    计算整体 R²、MAE、RMSE、PearsonR 指标

    参数：
        y_true: numpy array，真实值
        y_pred: numpy array，预测值

    返回：
        字典，包含所有评估指标
    """
    mae = mean_absolute_error(y_true, y_pred)
    rmse = root_mean_squared_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    pearson_r, _ = pearsonr(y_true, y_pred)

    print("✅ 总体评估指标：")
    print(f"MAE      = {mae:.4f}")
    print(f"RMSE     = {rmse:.4f}")
    print(f"R²       = {r2:.4f}")
    print(f"PearsonR = {pearson_r:.4f}")

    return {
        'MAE': mae,
        'RMSE': rmse,
        'R2': r2,
        'PearsonR': pearson_r
    }


```

#BERT-DNN

```python
errors=[]
all_val =  []
all_pred = []
for i in range(10):
    val = np.load(f'../k_folds_model/mlp_predictions/fold_{i+1}_y_true.npy')
    preds = np.load(f'../k_folds_model/mlp_predictions/fold_{i+1}_y_pred.npy')
    error = np.maximum(val, preds) / np.minimum(val, preds)
    
    errors.append(error)
    all_val.append(val)
    all_pred.append(preds)
errors = np.concatenate(errors)

all_val = np.concatenate(all_val)
all_pred = np.concatenate(all_pred)

evaluate_regression_metrics(all_val, all_pred)
```
