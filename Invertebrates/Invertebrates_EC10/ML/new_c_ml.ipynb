```python
import numpy as np
from sklearn.metrics import mean_absolute_error, r2_score, mean_squared_error
from scipy.stats import pearsonr, t

def root_mean_squared_error(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

def evaluate_single_fold(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = root_mean_squared_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    pearson_r, _ = pearsonr(y_true, y_pred)
    return mae, rmse, r2, pearson_r
    
# 计算均值和95%置信区间
def mean_ci(metric_list):
    mean = np.mean(metric_list)
    std = np.std(metric_list, ddof=1)
    ci = t.ppf(0.975, df=len(metric_list)-1) * std / np.sqrt(len(metric_list))
    return mean, ci
```

#LGB

```python
# 初始化容器
errors = []
all_val = []
all_pred = []

# 每折评估指标
mae_list, rmse_list, r2_list, pearson_list = [], [], [], []

# 遍历 10 折
for i in range(10):
    val = np.load(f'./lgbm_cv_results/fold_{i}_y_true.npy')
    preds = np.load(f'./lgbm_cv_results/fold_{i}_y_pred.npy')
    
    # 错误倍数
    error = np.maximum(val, preds) / np.minimum(val, preds)
    errors.append(error)
    
    all_val.append(val)
    all_pred.append(preds)
    
    # 每折指标
    mae, rmse, r2, pearson_r = evaluate_single_fold(val, preds)
    mae_list.append(mae)
    rmse_list.append(rmse)
    r2_list.append(r2)
    pearson_list.append(pearson_r)

# 拼接
errors = np.concatenate(errors)
all_val = np.concatenate(all_val)
all_pred = np.concatenate(all_pred)

# 汇总指标 ± CI
mae_mean, mae_ci = mean_ci(mae_list)
rmse_mean, rmse_ci = mean_ci(rmse_list)
r2_mean, r2_ci = mean_ci(r2_list)
pearson_mean, pearson_ci = mean_ci(pearson_list)

print("📊 LGB 每折评估指标均值 ± 95%置信区间：")
print(f"MAE      = {mae_mean:.3f} ± {mae_ci:.3f}")
print(f"RMSE     = {rmse_mean:.3f} ± {rmse_ci:.3f}")
print(f"R²       = {r2_mean:.3f} ± {r2_ci:.3f}")
print(f"PearsonR = {pearson_mean:.3f} ± {pearson_ci:.3f}")
```
